```python
import time
import pandas as pd
import polars as pl
```

```python
# making a Pandas dataframe
df_pd = pd.read_csv('data/person_id_map.csv', sep=",")
df_pd.info(show_counts=True)
```

```python
# making a Polars dataframe
df_pl = pl.read_csv('data/person_id_map.csv', separator=",", null_values=['null', '']) #
df_pl
### NOTE: data types not exactly the same
```

```python
# comparing shapes
print(f"df_pd.shape = {df_pd.shape}")
print(f"df_pl.shape = {df_pl.shape}")
```

```python
# using melt (Pandas)
start_time = time.time()
pd.melt(df_pd, id_vars='Watchmode ID', var_name='variable', value_name='value')
delta_time_pd = time.time() - start_time
delta_time_pd
```

```python
# using unpivot (Polars)
start_time = time.time()
df_pl.unpivot(index="Watchmode ID")
delta_time_pl = time.time() - start_time
delta_time_pl
```

```python
# reduction of X percent
percent_reduced = ((delta_time_pd - delta_time_pl) / delta_time_pd)*100
print(f"using Polars reduced time by {percent_reduced:.0f}%")
```
